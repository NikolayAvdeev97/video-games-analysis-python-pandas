  Причины: Вероятнее всего данные игры не были рейтированы.
  Решение: Так как данный столбец не потребуется для дальнейшего анализа, то можно оставить пропуски как есть.

- Обработаем пропущенные значения. Для каждого случая выберем оптимальный вариант.

```python
#Сохраним изначальное кол-во строк в датафрейме и удалим строки с пропусками в столбцах `name`, `genre`, `year_of_release`
initial_row_count1 = ng.shape[0]
ng = ng.dropna(subset=['name', 'genre', 'year_of_release'])
```

```python
#В столбцах `eu_sales` и `jp_sales` заменим пропуски на среднее значение в зависимости от названия платформы и года выхода игры
def mean_group_sales(row):
    if pd.isna(row[['eu_sales', 'jp_sales']]).any():
        group = ng[
            (ng['platform'] == row['platform']) &
            (ng['year_of_release'] == row['year_of_release'])
        ]
        mean_eu = group['eu_sales'].mean()
        mean_jp = group['jp_sales'].mean()
        if pd.isna(row['eu_sales']):
            row['eu_sales'] = mean_eu
        if pd.isna(row['jp_sales']):
            row['jp_sales'] = mean_jp
    return row

ng = ng.apply(mean_group_sales, axis=1)
```

```python
#В столбцах `critic_score` и `user_score` заменим пропуски на значение-индикатор (-1)
ng[['critic_score', 'user_score']] = ng[['critic_score', 'user_score']].fillna(-1)
```

```python
#Проверим кол-во строк после обработки пропусков
final_row_count = ng.shape[0]
display(f"Количество строк после обработки пропусков: {final_row_count}")
```

```python
#Проверим столбцы на оставшиеся пропуски
display(ng.isna().sum())
```

- Исходя из данных, видим, что замена и удаление пропусков прошло успешно. Остались только пропущенные знаяения
в столбце `rating`, но т.к. данный столбец не входит анализ, то было принято решение оставить его без изменений.

Перед нами стоит цель изучить историю продаж игр в начале XXI века в период с 2000 по 2013 год включительно. Отберем данные по этому показателю. Сохраним новый срез данных в отдельном датафрейме `ng_xxi`.

```python
#Произведем фильтрацию по столбцу `year_of_release` с помощью логической маски
ng_xxi = ng[(ng['year_of_release'] >= 2000) & (ng['year_of_release'] <= 2013)].copy()
display(ng_xxi.head())
```

## 4. Категоризация данных
    
Проведем категоризацию данных:
- Разделим все игры по оценкам пользователей и выделим такие категории: высокая оценка (от 8 до 10 включительно), средняя оценка (от 3 до 8, не включая правую границу интервала) и низкая оценка (от 0 до 3, не включая правую границу интервала).

```python
#Категоризуем игры по оценкам пользователей
user_grade_groups = []

for grade in ng_xxi['user_score']:
    if 8 <= grade <= 10:
        user_grade_groups.append("высокая оценка")
    elif 3 <= grade < 8:
        user_grade_groups.append("средняя оценка")
    elif 0 <= grade < 3:
        user_grade_groups.append("низкая оценка")
    else:
        user_grade_groups.append("Нет оценки")
        
ng_xxi['user_grade_group'] = user_grade_groups
```

- Разделим все игры по оценкам критиков и выделим такие категории: высокая оценка (от 80 до 100 включительно), средняя оценка (от 30 до 80, не включая правую границу интервала) и низкая оценка (от 0 до 30, не включая правую границу интервала).

```python
#Категоризуем игры по оценкам критиков
critic_grade_groups = []

for grade in ng_xxi['critic_score']:
    if 80 <= grade <= 100:
        critic_grade_groups.append("высокая оценка")
    elif 30 <= grade < 80:
        critic_grade_groups.append("средняя оценка")
    elif 0 <= grade < 30:
        critic_grade_groups.append("низкая оценка")
    else:
        critic_grade_groups.append("Нет оценки")
        
ng_xxi['critic_grade_group'] = critic_grade_groups
display(ng_xxi.head())
```

- После категоризации данных проверим результат: сгруппируем данные по выделенным категориям и посчитаем количество игр в каждой категории.

```python
#Произведем группировку по столбцам `user_grade_group` и `critic_grade_group` и подсчитаем кол-во игр по выделенным 
#категориям
count_games_category = ng_xxi.groupby(['critic_grade_group', 'user_grade_group'])['name'].count()
display(count_games_category)
```

```python
#Произведем подсчет игр по каждой платформе и выделим 7 из них с наибольшим количеством
count_games_platform = ng_xxi['platform'].value_counts()
display(count_games_platform.head(7))
```
